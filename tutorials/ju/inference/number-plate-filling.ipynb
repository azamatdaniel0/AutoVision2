```python
import os
import sys

current_dir = os.path.dirname("./")
nomeroff_net_dir = os.path.join(current_dir, "../../../")
sys.path.append(nomeroff_net_dir)

import glob
import matplotlib.pyplot as plt
```

```python
from nomeroff_net import pipeline
```

```python
number_plate_bbox_filling = pipeline("number_plate_filling", image_loader="opencv")
```

```python
imgs = glob.glob(os.path.join(os.path.join(nomeroff_net_dir, './data/examples/oneline_images/*')))
```

```python
images = number_plate_bbox_filling(imgs)
```

```python
%matplotlib inline
plt.rcParams["figure.figsize"] = (20,10)

for image in images:
    plt.imshow(image)
    plt.show()
```
